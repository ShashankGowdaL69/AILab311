```python
import heapq

print("ShashankGowdaL 1BM23CS311")
goal_state = [[1, 2, 3],
              [4, 5, 6],
              [7, 8, 0]]

def manhattan_distance(state, goal):
    distance = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0:
                goal_x, goal_y = divmod(goal_state[i][j] - 1, 3)
                distance += abs(i - goal_x) + abs(j - goal_y)
    return distance

def find_empty(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return (i, j)

def get_neighbors(state):
    neighbors = []
    x, y = find_empty(state)
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for dx, dy in directions:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [row[:] for row in state]
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append(new_state)
    return neighbors

def a_star(start_state, goal_state):
    open_list = []
    heapq.heappush(open_list, (manhattan_distance(start_state, goal_state), 0, start_state, []))
    closed_list = set()

    while open_list:
        _, g, current_state, path = heapq.heappop(open_list)
        if current_state == goal_state:
            return path + [current_state]

        closed_list.add(tuple(map(tuple, current_state)))

        for neighbor in get_neighbors(current_state):
            if tuple(map(tuple, neighbor)) not in closed_list:
                f = g + 1 + manhattan_distance(neighbor, goal_state)
                heapq.heappush(open_list, (f, g + 1, neighbor, path + [current_state]))

    return None

start_state = [[1, 2, 3],
               [4, 5, 6],
               [7, 8, 0]]

solution = a_star(start_state, goal_state)
for step in solution:
    print(step)
```
