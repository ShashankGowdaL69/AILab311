```python
import random


winning_combinations = [
    # Horizontal
    [1, 2, 3], [2, 3, 4], [5, 6, 7], [6, 7, 8], [7, 8, 9],
    # Vertical
    [1, 2, 5], [2, 3, 6], [3, 4, 7], [4, 7, 9], [5, 6, 8],
    # Diagonal
    [1, 3, 6], [2, 5, 7], [3, 6, 8], [4, 7, 8], [2, 4, 7], [3, 5, 7]
]

def print_board(board):
    print("\nCurrent board:")
    print(f"    {board[1]}")
    print(f"   {board[2]} {board[3]} {board[4]}")
    print(f"  {board[5]} {board[6]} {board[7]} {board[8]} {board[9]}")
    print()

def check_winner(board, symbol):
    for combo in winning_combinations:
        if all(board[pos] == symbol for pos in combo):
            return True
    return False

def available_moves(board):
    return [pos for pos in range(1, 10) if board[pos] == " "]

def computer_move(board):
    moves = available_moves(board)
    return random.choice(moves)

def play_game():
    board = [" "] * 10
    current_player = "user"
    user_symbol = "X"
    computer_symbol = "O"


    print_board(board)

    for turn in range(9):
        if current_player == "user":
            try:
                move = int(input("Enter your move (1-9): "))
                if move not in available_moves(board):
                    print("Invalid move. Try again.")
                    continue
            except ValueError:
                print("Please enter a number between 1 and 9.")
                continue
            board[move] = user_symbol
            if check_winner(board, user_symbol):
                print_board(board)
                print("You win!")
                return
            current_player = "computer"
        else:
            move = computer_move(board)
            board[move] = computer_symbol
            print(f"Computer chooses {move}")
            if check_winner(board, computer_symbol):
                print_board(board)
                print("Computer wins!")
                return
            current_player = "user"

        print_board(board)

    print("It's a draw!")

if __name__ == "__main__":
    play_game()
```
